## Conclusion

- The Sensor Readings dataset contains **50,000 records and 9 columns** related to vaccine shipment monitoring. The initial data understanding process was performed to identify missing values, duplicate records, data distribution, and potential data-quality issues.

- The analysis found **no missing values** and **no duplicate rows**. The `reading_id` values are unique, while multiple readings are associated with the same shipment and container, which is expected for sensor data.

- The sensor values were also examined. Humidity, vibration, latitude, and longitude showed reasonable ranges. However, **2,894 temperature readings were above 8°C**, with temperatures reaching a maximum of **21.55°C**. These records should not be removed automatically because they may represent genuine **cold-chain temperature excursions**.

- Therefore, the high-temperature readings will be **identified and flagged during the data-cleaning stage**, while preserving the original sensor values. Further validation and cleaning will be performed in the next notebook.
