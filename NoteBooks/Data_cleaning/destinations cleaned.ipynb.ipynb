```python
import pandas as pd
import numpy as np

# 1. Load the raw dataset
file_path = 'destinations_demand_raw.xls'
df = pd.read_csv(file_path)
print(f"Initial dataset shape: {df.shape}")
df.head(10)
```

```python
print("--- Data Info ---")
print(df.info())
```

```python
print("\n--- Missing Values ---")
print(df.isnull().sum())
```

```python
df_clean = df.copy()
```

```python
initial_len = len(df_clean)
df_clean = df_clean.drop_duplicates()
print(f"Dropped {initial_len - len(df_clean)} duplicate rows.")
```

```python
df_clean['destination'] = df_clean['destination'].astype(str).str.strip().str.title()
df_clean['priority'] = df_clean['priority'].astype(str).str.strip().str.capitalize()
df_clean['demand'] = df_clean['demand'].astype(str).str.strip().str.title()
df_clean.loc[df_clean['demand'].isin(['Nan', 'None']), 'demand'] = np.nan
```

```python
high_df = df_clean[df_clean['priority'] == 'High']
median_high_qty = high_df[high_df['required_quantity'] > 0]['required_quantity'].median()
df_clean.loc[df_clean['required_quantity'] < 0, 'required_quantity'] = median_high_qty
```

```python
df_clean.loc[df_clean['demand'].isnull() & (df_clean['required_quantity'] > 1000), 'demand'] = 'Very High'
```

```python
very_high_df = df_clean[df_clean['demand'] == 'Very High']
median_very_high_qty = very_high_df['required_quantity'].median()
df_clean.loc[df_clean['required_quantity'].isnull(), 'required_quantity'] = median_very_high_qty

print("\n--- Cleaned DataFrame Summary ---")
print(df_clean.info())
print(df_clean.isnull().sum())
```

```python
print("Cleaned Data Preview:")
df_clean.reset_index(drop=True, inplace=True)
df_clean
```

```python
output_csv_path = 'destinations_demand_cleaned.csv'
output_excel_path = 'destinations_demand_cleaned.xlsx'

df_clean.to_csv(output_csv_path, index=False)
df_clean.to_excel(output_excel_path, index=False)

print(f"Successfully exported cleaned dataset to '{output_csv_path}' and '{output_excel_path}'.")
```
